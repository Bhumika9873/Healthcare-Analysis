```python
# Data cleaning
df = df.drop_duplicates().copy()
```

```python
df.duplicated().sum()
```

```python
df["Name"] = df["Name"].str.strip().str.title()
```

```python
df["Gender"] = df["Gender"].str.strip().str.title()
df["Blood Type"] = df["Blood Type"].str.strip().str.upper()
df["Medical Condition"] = df["Medical Condition"].str.strip().str.title()
df["Insurance Provider"] = df["Insurance Provider"].str.strip().str.title()
df["Admission Type"] = df["Admission Type"].str.strip().str.title()
df["Medication"] = df["Medication"].str.strip().str.title()
df["Test Results"] = df["Test Results"].str.strip().str.title()
```

```python
df["Name"].head(10)
```

```python
df["Gender"].value_counts()
```

```python
# Date conversion
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df[["Date of Admission", "Discharge Date"]].head()
```

```python
# Feature engineering
df["Length of Stay"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days
```

```python
df[["Date of Admission", "Discharge Date", "Length of Stay"]].head()
```

```python
bins = [0, 18, 35, 50, 65, 100]
labels = ["0-18", "19-35", "36-50", "51-65", "66+"]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
```

```python
df[["Age", "Age Group"]].head(10)
```

```python
df = df.reset_index(drop=True)
```

```python
df.head()
```

```python
# Export cleaned dataset
df.to_csv(r"D:\HealthcareProject\data\Cleaned_Healthcare.csv", index=False)
```
